## Statistical Tests and Correlation Analysis

### ANOVA Test for Categorical Features vs. `depression_label`

We will perform an ANOVA test to check if there is a statistically significant difference in the mean `depression_label` across different categories within each categorical feature. A low p-value (typically < 0.05) suggests a significant difference.

```python
from scipy.stats import f_oneway

cat_cols = df.select_dtypes(include='object').columns.tolist()

print('ANOVA Test Results for Categorical Features vs. depression_label:')
print('--------------------------------------------------------------')

for col in cat_cols:
    if col != 'depression_label': # Ensure we don't test the target against itself if it somehow ends up in cat_cols
        groups = df[col].unique()
        data_for_anova = [df[df[col] == g]['depression_label'] for g in groups]

        # Perform ANOVA test
        statistic, pvalue = f_oneway(*data_for_anova)

        print(f'Feature: {col}')
        print(f'  F-statistic: {statistic:.4f}')
        print(f'  P-value: {pvalue:.4f}')
        if pvalue < 0.05:
            print('  Interpretation: Significant difference in depression_label across categories (Reject H0)')
        else:
            print('  Interpretation: No significant difference in depression_label across categories (Fail to Reject H0)')
        print('--------------------------------------------------------------')
```

```python
from sklearn.preprocessing import OrdinalEncoder,OneHotEncoder,StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
```

```python
df.columns
```

```python
skewd_cols=['physical_activity','anxiety_level','addiction_level']
ordinal_encode_cols=['social_interaction_level']
one_hot_cols=['gender','platform_usage']
scaler_cols=['age','daily_social_media_hours','sleep_hours','screen_time_before_sleep','academic_performance','stress_level']

feature_cols=skewd_cols+ordinal_encode_cols+one_hot_cols+scaler_cols
target_col=['depression_label']
```

```python
from imblearn.over_sampling import SMOTE
```

```python
from sklearn.preprocessing import FunctionTransformer
```

```python
df['social_interaction_level'].unique()
```

```python
skewd_pipeline=Pipeline(steps=[
    ('skew',FunctionTransformer(np.log1p)),
    ('scalar',StandardScaler())
])

ordinal_pipeline=Pipeline(steps=[
    ('ordinal',OrdinalEncoder(categories=[['low', 'high', 'medium']]))
])

one_hot_pipelne=Pipeline(steps=[
    ('one_hot',OneHotEncoder(handle_unknown='ignore'))
])

scaler_pipeline=Pipeline(steps=[
    ('scalar',StandardScaler())
])



preprocessor=ColumnTransformer(transformers=[
    ('skewd',skewd_pipeline,skewd_cols),
    ('ordinal',ordinal_pipeline,ordinal_encode_cols),
    ('one_hot',one_hot_pipelne,one_hot_cols),
    ('scaler',scaler_pipeline,scaler_cols)
])
```

```python
x=df.drop('depression_label',axis=1)
y=df['depression_label']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,test_size=0.2)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
smote = SMOTE(random_state=42)

# Apply preprocessor first
x_train_processed = preprocessor.fit_transform(x_train, y_train)
x_test_processed = preprocessor.transform(x_test)

# Apply SMOTE to the processed training data
x_train_resampled, y_train_resampled = smote.fit_resample(x_train_processed, y_train)

# Define and fit the logistic regression model with resampled data
logistic_model = LogisticRegression(solver='liblinear', random_state=42) # Added solver for convergence
logistic_model.fit(x_train_resampled, y_train_resampled)

# Now you can make predictions and evaluate the model
y_pred = logistic_model.predict(x_test_processed)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
knn_pipeline=Pipeline(steps=[
    ('preprocessor',preprocessor),
    ('knn',GaussianNB())
])
```

```python
knn_pipeline.fit(x_train,y_train)
```

```python
knn_pred=knn_pipeline.predict(x_test)
knn_train_pred=knn_pipeline.predict(x_train)
```

```python
print("Accuracy:", accuracy_score(y_test, knn_pred))
print("\nClassification Report:\n", classification_report(y_test, knn_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, knn_pred))
print("\nTrain Classification Report:\n", classification_report(y_train, knn_train_pred))
print("\nTrain Confusion Matrix:\n", confusion_matrix(y_train, knn_train_pred))
print(f'Training Score: {knn_pipeline.score(x_train,y_train)}')
print(f'Testing Score: {knn_pipeline.score(x_test,y_test)}')
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf_pipeline=Pipeline(steps=[
    ('preprocessor',preprocessor),
    ('rf',RandomForestClassifier(random_state=42))
])

rf_pipeline.fit(x_train,y_train)
rf_pred=rf_pipeline.predict(x_test)
rf_train_pred=rf_pipeline.predict(x_train)

print("Accuracy:", accuracy_score(y_test, rf_pred))
print("\nClassification Report:\n", classification_report(y_test, rf_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, rf_pred))
print("\nTrain Classification Report:\n", classification_report(y_train, rf_train_pred))
print("\nTrain Confusion Matrix:\n", confusion_matrix(y_train, rf_train_pred))
print(f'Training Score: {rf_pipeline.score(x_train,y_train)}')
print(f'Testing Score: {rf_pipeline.score(x_test,y_test)}')
```

```python
from sklearn.model_selection import GridSearchCV

rf_tuned_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('randomforest', RandomForestClassifier(random_state=42))
])

param_grid = {
    'randomforest__n_estimators': [100, 200, 300],
    'randomforest__max_depth': [10, 20, 30, None],
    'randomforest__min_samples_split': [2, 5, 10],
    'randomforest__min_samples_leaf': [1, 2, 4],
    'randomforest__class_weight': ['balanced', None]
}


print("Starting GridSearchCV for Random Forest...")

x_train_processed_for_tuning = preprocessor.fit_transform(x_train, y_train)
x_train_resampled_for_tuning, y_train_resampled_for_tuning = smote.fit_resample(x_train_processed_for_tuning, y_train)


rf_model_for_tuning = RandomForestClassifier(random_state=42)


param_grid_direct = {
    'n_estimators': [100, 200, 300],
    'max_depth': [10, 20, 30, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'class_weight': ['balanced', None]
}

grid_search_rf = GridSearchCV(estimator=rf_model_for_tuning, param_grid=param_grid_direct, cv=5, scoring='f1', n_jobs=-1, verbose=2)
grid_search_rf.fit(x_train_resampled_for_tuning, y_train_resampled_for_tuning)

print("Best parameters found: ", grid_search_rf.best_params_)
print("Best F1-score found: ", grid_search_rf.best_score_)


best_rf_model = grid_search_rf.best_estimator_


x_test_processed_for_eval = preprocessor.transform(x_test)
y_pred_tuned_rf = best_rf_model.predict(x_test_processed_for_eval)

print("\n--- Tuned Random Forest Model Evaluation ---")
print("Accuracy:", accuracy_score(y_test, y_pred_tuned_rf))
print("\nClassification Report:\n", classification_report(y_test, y_pred_tuned_rf))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred_tuned_rf))
```

```python
from xgboost import XGBClassifier

xgb_model_for_tuning = XGBClassifier(random_state=42, use_label_encoder=False, eval_metric='logloss')

param_grid_xgb = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 5, 7],
    'learning_rate': [0.01, 0.1, 0.2],
    'subsample': [0.7, 0.9, 1.0],
    'colsample_bytree': [0.7, 0.9, 1.0]
}

print("Starting GridSearchCV for XGBoost...")

grid_search_xgb = GridSearchCV(
    estimator=xgb_model_for_tuning,
    param_grid=param_grid_xgb,
    cv=5,
    scoring='f1',
    n_jobs=-1,
    verbose=2
)


grid_search_xgb.fit(x_train_resampled_for_tuning, y_train_resampled_for_tuning)

print("Best parameters found for XGBoost: ", grid_search_xgb.best_params_)
print("Best F1-score found for XGBoost: ", grid_search_xgb.best_score_)

best_xgb_model = grid_search_xgb.best_estimator_

y_pred_tuned_xgb = best_xgb_model.predict(x_test_processed_for_eval)

print("\n--- Tuned XGBoost Model Evaluation ---")
print("Accuracy:", accuracy_score(y_test, y_pred_tuned_xgb))
print("\nClassification Report:\n", classification_report(y_test, y_pred_tuned_xgb))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred_tuned_xgb))
```

### Model Comparison and Recommendation

Based on the comparison table above, here's an analysis of the models, with a particular focus on the minority class (`depression_label = 1`) due to the class imbalance:

*   **Tuned XGBoost:** This model stands out as the best performer. It achieves the highest F1-Score for the minority class (0.91) with excellent precision (1.00) and recall (0.83). Its overall accuracy is also very high (0.9958).

*   **Gaussian Naive Bayes:** This model also performed very well, particularly for the minority class, with an F1-Score of 0.80. It has perfect precision (1.00) for class 1, indicating that when it predicts a positive case, it's very likely correct. However, its recall for class 1 (0.67) is slightly lower than XGBoost.

*   **Tuned Random Forest:** While improving significantly over the untuned version, the Tuned Random Forest model achieved an F1-Score of 0.67 for the minority class. Its recall for class 1 (0.50) is lower than both XGBoost and Gaussian Naive Bayes, meaning it misses more actual positive cases.

*   **Logistic Regression:** This model showed the weakest performance for the minority class, with an F1-Score of 0.43 and a low precision of 0.29, despite a decent recall of 0.83. This suggests it has a high false positive rate for the minority class.

**Recommendation:**

Considering the goal of predicting `depression_label`, where correctly identifying the minority class (individuals with depression) is crucial, the **Tuned XGBoost model** is the most robust and recommended model among those evaluated. It provides the best balance of precision and recall for the minority class, leading to the highest F1-score, which is often the preferred metric for imbalanced classification problems.

```python
from sklearn.pipeline import Pipeline
import joblib

# Assuming `preprocessor` and `best_xgb_model` are already defined and fitted.
# The `preprocessor` was defined in cell '4Y5sviuJb0ZB'.
# The `best_xgb_model` was obtained from 'grid_search_xgb.best_estimator_' in cell '610c22aa'.

# Create a combined pipeline
final_model_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', best_xgb_model)
])

# Fit the combined pipeline using the resampled training data.
# Note: When fitting the full pipeline, you typically fit it on the raw training data (x_train, y_train)
# and the preprocessor handles the transformation internally. However, since we already
# handled SMOTE and fitted the preprocessor separately, we need to re-evaluate how to train this final pipeline.
# A common approach is to refit the entire pipeline on the processed and resampled data if SMOTE is part of the pipeline.
# But since SMOTE is external to the preprocessor, we need to ensure the final pipeline is trained correctly.

# For simplicity and to use the best_xgb_model that was already tuned on x_train_resampled_for_tuning,
# we will assume the 'preprocessor' is already fitted and we are just combining the *fitted* components.
# However, for a truly robust deployment, the SMOTE step should be either:
# 1. Removed for prediction (as it's only for training).
# 2. Integrated into a pipeline that's only applied during fitting, not transforming new data.
# For prediction, we only need the preprocessor and the classifier.

# To train this 'final_model_pipeline' from scratch:
# 1. Remove SMOTE (it's a training-only step)
# 2. Potentially integrate SMOTE into a new Pipeline step that only runs during .fit()
# For this demonstration, we'll assume the preprocessor is already fitted and the best_xgb_model is ready.

# A more common and robust way to use SMOTE in a pipeline for final model saving is with imblearn's Pipeline:
from imblearn.pipeline import Pipeline as ImbPipeline

final_pipeline_with_smote = ImbPipeline(steps=[
    ('preprocessor', preprocessor),
    ('smote', SMOTE(random_state=42)),
    ('classifier', best_xgb_model) # Use the best_xgb_model obtained from GridSearchCV
])

# We fit this *new* pipeline on the original x_train, y_train.
# This will re-fit the preprocessor and then apply SMOTE.
# This is the most correct way if you want to save a pipeline that includes SMOTE for future training/retraining.
# For immediate deployment where SMOTE is not needed for new predictions, you save the preprocessor and the classifier.

# For prediction, we don't need SMOTE. The best_xgb_model was tuned on already resampled data.
# So, the direct pipeline for prediction is:
prediction_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', best_xgb_model)
])

# Let's save this prediction_pipeline which is what you'd use in FastAPI for new inferences.
filename = 'xgboost_mental_health_prediction_pipeline.pkl'
joblib.dump(prediction_pipeline, filename)

print(f"Combined prediction pipeline saved to {filename}")
```
